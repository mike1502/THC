# Diccionarios
[regresar al contenido](#c1)

En Python, un diccionario (dict) es una estructura de datos que almacena elementos en pares de clave-valor (key-value). Las claves deben ser únicas e inmutables (cadenas, números o tuplas), mientras que los valores pueden ser de cualquier tipo. Los diccionarios son muy útiles para representar colecciones de datos en las que necesitas acceder a los elementos por su clave en lugar de por su posición, como lo harías en una lista.

# Declaración
[regresar al contenido](#c1)

Existen dos formas principales para declarar diccionarios:

1. Usando llaves {} (La forma más común y rápida)
Es el estándar en Python por su legibilidad y mejor rendimiento. Los pares se separan por comas y la clave del valor mediante dos puntos (:).

```python
# Diccionario con datos
usuario = {
    "nombre": "Ana",
    "edad": 28,
    "activo": True,
    "lenguajes": ["Python", "SQL"]
}

# Diccionario vacío
vacio = {}

print(usuario)
```

```python
print(vacio)
```

2. Usando el constructor dict()
Es útil cuando las claves son cadenas de texto simples sin espacios ni caracteres especiales, ya que evita tener que escribir comillas para las claves, o cuando necesitas transformar otra estructura de datos en un diccionario.

```python
# Pasando argumentos con nombre (keyword arguments)
usuario = dict(nombre="Ana", edad=28, activo=True)

# A partir de una lista de tuplas (pares clave-valor)
pares = [("nombre", "Ana"), ("edad", 28)]
usuario_desde_tuplas = dict(pares)

# Diccionario vacío
vacio = dict()
```

# Método .keys() y método .values()
[regresar al contenido](#c1)

Devuelve una vista con todas las claves del diccionario. Es muy útil para verificar si una clave existe o para iterar sobre ellas. **nombre_diccionario.keys()**

Por otro lado, values() devuelve una vista con todos los valores almacenados, sin sus claves.

```python
config = {"tema": "oscuro", "idioma": "es", "notificaciones": True}

claves = config.keys()     # dict_keys(['tema', 'idioma', 'notificaciones'])
valores = config.values()  # dict_values(['oscuro', 'es', True])

# Si agregamos un dato nuevo, las vistas se actualizan solas:
config["volumen"] = 80
print(claves)  # dict_keys(['tema', 'idioma', 'notificaciones', 'volumen'])
```

# Método items y get

items(): Devuelve una vista de tuplas (clave, valor). Es el método estándar para recorrer un diccionario completo en un bucle for.

get(clave, valor_por_defecto): Obtiene el valor de una clave sin lanzar un error (KeyError) si no existe. Si no se encuentra, devuelve None o el valor por defecto que indiques.

```python
# Ejemplo de items() y get()
for k, v in config.items():
    print(f"{k}: {v}")

# Acceso seguro (evita que el programa se detenga si 'fuente' no existe)
fuente = config.get("fuente", "Arial")  # Devuelve 'Arial'
```

# Método de acceso a elementos
[regresar al contenido](#c1)

Existen dos formas principales para acceder al valor de un elemento en un diccionario:

1. Acceso directo con corchetes (nombre_diccionario['clave']): Devuelve el valor asociado, pero detiene el programa con un KeyError si la clave no existe.

2. Acceso seguro con el método .get() (nombre_diccionario.get('clave')): Devuelve el valor si la clave existe, o None (sin lanzar error) si no la encuentra.

```python
# Creamos un diccionario sencillo
producto = {
    "nombre": "Teclado mecánico",
    "precio": 850
}

# 1. Cuando la clave SÍ existe: ambos funcionan igual
print(producto["nombre"])      # Salida: Teclado mecánico
print(producto.get("nombre"))  # Salida: Teclado mecánico

# 2. Cuando la clave NO existe: aquí está la diferencia
print(producto.get("descuento"))      # Salida: None (el programa continúa sin fallar)
print(producto.get("descuento", 0))   # Salida: 0 (devuelve el valor por defecto que indiques)

# Si descomentas la siguiente línea, el código se detendrá con un error:
# print(producto["descuento"])        # Lanza KeyError: 'descuento'
```

Usa diccionario['clave'] cuando estés 100% seguro de que la clave existe (o cuando quieras que el programa falle intencionalmente si falta ese dato).

Usa diccionario.get('clave') cuando la clave sea opcional o provenga de datos externos donde no tienes garantía de que esté presente.

# Métodos para diccionarios


| Método | Sintaxis / Parámetros | Descripción | Valor de retorno |
| :--- | :--- | :--- | :--- |
| **`keys()`** | `d.keys()` | Devuelve una vista dinámica con todas las claves del diccionario. | Objeto `dict_keys` |
| **`values()`** | `d.values()` | Devuelve una vista dinámica con todos los valores almacenados. | Objeto `dict_values` |
| **`items()`** | `d.items()` | Devuelve una vista dinámica con los pares `(clave, valor)` en forma de tuplas. | Objeto `dict_items` |
| **`get()`** | `d.get(clave, defecto=None)` | Obtiene el valor de una clave sin lanzar `KeyError` si no existe. | El valor asociado, o `defecto` (`None` si no se indica). |
| **`update()`** | `d.update(otro_dict)` | Fusiona otro diccionario o iterable de pares en el actual, sobrescribiendo claves existentes. | `None` (modifica el diccionario *in-place*). |
| **`setdefault()`** | `d.setdefault(clave, defecto=None)` | Busca una clave; si no existe, la inserta con el valor `defecto`. | El valor existente o el nuevo valor insertado. |
| **`pop()`** | `d.pop(clave, defecto)` | Elimina la clave indicada. Lanza `KeyError` si no existe y no se especifica `defecto`. | El valor de la clave eliminada, o `defecto`. |
| **`popitem()`** | `d.popitem()` | Elimina el **último** par `(clave, valor)` insertado (orden LIFO). Lanza `KeyError` si está vacío. | Tupla `(clave, valor)` eliminada. |
| **`clear()`** | `d.clear()` | Elimina todos los elementos del diccionario, dejándolo vacío (`{}`). | `None` (modifica el diccionario *in-place*). |
| **`copy()`** | `d.copy()` | Crea una copia superficial (*shallow copy*) del diccionario. | Un nuevo objeto `dict`. |
| **`fromkeys()`** | `dict.fromkeys(iterable, valor=None)` | Método de clase que crea un diccionario nuevo usando el iterable como claves con un mismo valor inicial. | Un nuevo objeto `dict`. |
