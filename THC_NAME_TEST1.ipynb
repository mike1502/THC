```python
import math

a = 6378137.0
c = 6356752.314245
r = 6371000.0

excentricidad = 1 - ((c**(2)) / (a**(2)))
e = math.sqrt(excentricidad)

ec1 = 1-excentricidad
ec2 = (ec1/(e))*math.atanh(e)

s_obl = (2*math.pi*(a**2))*(1+ec2)
print(f"superficie del esferoide oblato: (s_obl)")
se = 4*math.pi*(r**2)
print("superficie de la superficie de la esfera",se)
dif=math.fabs((se-s_obl)/se)*100

print(f"la diferencia es del orden de {dif:.10f}%")
```
