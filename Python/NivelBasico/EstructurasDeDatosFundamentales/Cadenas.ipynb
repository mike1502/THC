# Cadenas
[regresar al contenido](#c1)

Las cadenas, también conocidas como strings en programación, son secuencias de caracteres (números, marcas de puntuación, letras, etc.). En Python, se pueden definir cadenas utilizando comillas simples (') o comillas dobles ("). Se pueden utilizar para representar información textual y se almacenan en variables.

## Reemplazar elementos en una cadena
[regresar al contenido](#c1)

Si desea borrar el valor de la lista en un índice específico se hace de la siguiente manera

```python
# Definir una cadena
cadena1 = "Hola, mundo!"
cadena2 = 'Python es genial'

# Reemplazar parte de una cadena
nueva_cadena = cadena2.replace("genial", "increíble")
print(nueva_cadena)  # Salida: Python es increíble
```

## Método Split y slicing
[regresar al contenido](#c1)

```python
# Separar una cadena en una lista de palabras
palabras = cadena2.split()  # Divide la cadena en palabras
print(palabras)  # Salida: ['Python', 'es', 'genial']
```

El slicing se hace cuando quieres tomar solamente una sección de la cadena, se usa de la siguiente manera **nombre_Cadena[indice_Inicial : indice_Final]**

```python
# Subcadenas
subcadena = cadena1[6:11]  # Obtiene la subcadena "mundo"
print(subcadena)
```

## Striding (paso)


```python
subcadena_paso=cadena1[::2]
print(subcadena_paso)
```

# Pasar valores de una lista a una cadena

# Utilidad de las cadenas
[regresar al contenido](#c1)

```python
# Inicializar una lista de tareas vacía
lista_de_tareas = []

# Función para agregar una tarea a la lista
def agregar_tarea(tarea):
    lista_de_tareas.append(tarea)
    print("Tarea agregada:", tarea)

# Función para eliminar una tarea de la lista
def eliminar_tarea(tarea):
    if tarea in lista_de_tareas:
        lista_de_tareas.remove(tarea)
        print("Tarea eliminada:", tarea)
    else:
        print("La tarea no existe:", tarea)

# Función para mostrar todas las tareas
def mostrar_tareas():
    print("Tareas pendientes:")
    for tarea in lista_de_tareas:
        print("-", tarea)

# Agregar algunas tareas
agregar_tarea("Hacer compras")
agregar_tarea("Estudiar para el examen")
agregar_tarea("Ir al gimnasio")

# Mostrar las tareas actuales
mostrar_tareas()

# Eliminar una tarea
eliminar_tarea("Estudiar para el examen")

# Mostrar las tareas actualizadas
mostrar_tareas()
```

## Operaciones con cadenas (Tablas)

<a name='6'></a>
## Operaciones básicas

| Operación | Ejemplo | Resultado |
|---|---|---|
| Longitud | `len("Hola")` | `4` |
| Indexado | `"Hola"[1]` | `'o'` |
| Índice negativo | `"Hola"[-1]` | `'a'` |
| Slicing | `"abcdef"[1:4]` | `'bcd'` |
| Slicing invertido | `"abcd"[::-1]` | `'dcba'` |
| Pertenencia | `'a' in 'banana'` | `True` |
| Concatenar | `"py" + "thon"` | `'python'` |
| Repetir | `"ha"*3` | `'hahaha'` |
| Comparar | `"a" < "b"` | `True` |

## Cambio de mayúsculas/minúsculas

| Método | Efecto | Ejemplo | Resultado |
|---|---|---|---|
| `lower()` | a minúsculas | `"Hóla".lower()` | `'hóla'` |
| `upper()` | a MAYÚSCULAS | `"Hola".upper()` | `'HOLA'` |
| `capitalize()` | 1ª letra mayúscula | `"hola mundo".capitalize()` | `'Hola mundo'` |
| `title()` | Cada Palabra Capitalizada | `"hola mundo".title()` | `'Hola Mundo'` |
| `swapcase()` | Invierte mayúsculas/minúsculas | `"HoLa".swapcase()` | `'hOlA'` |
| `casefold()` | Minúsculas agresivas (comparación) | `"naïve".casefold()` | `'naive'` o similar |

## Búsqueda y conteo

| Método | Uso | Devuelve | Notas |
|---|---|---|---|
| `count(sub[, start[, end]])` | `"banana".count("a")` | `3` | Cuenta ocurrencias |
| `find(sub[, start[, end]])` | `"banana".find("na")` | `2` | 1er índice o `-1` |
| `rfind(sub)` | `"banana".rfind("na")` | `4` | Último índice o `-1` |
| `index(sub)` | `"banana".index("x")` | `ValueError` | Como `find` pero con error |
| `rindex(sub)` | `"banana".rindex("a")` | `5` | Como `rfind` pero con error |
| `startswith(pref[, start[, end]])` | `"python".startswith("py")` | `True` | |
| `endswith(suf[, start[, end]])` | `"data.csv".endswith(".csv")` | `True` | |

## Reemplazo y traducción

| Método | Uso | Resultado | Notas |
|---|---|---|---|
| `replace(viejo, nuevo[, max])` | `"hola hola".replace("hola","hey",1)` | `'hey hola'` | Reemplazo simple |
| `maketrans(...) + translate(tabla)` | `str.maketrans({'á':'a'})` + `"canción".translate(tabla)` | `'cancion'` | Múltiples reemplazos |

## Separar y unir

| Método | Uso | Resultado | Notas |
|---|---|---|---|
| `split(sep=None, maxsplit=-1)` | `"a b  c".split()` | `['a','b','c']` | Divide por espacios si `sep=None` |
| `rsplit(sep=None, maxsplit=-1)` | `"a,b,c".rsplit(",",1)` | `['a','b,c']` | Divide desde la derecha |
| `splitlines(keepends=False)` | `"a\nb".splitlines()` | `['a','b']` | Por saltos de línea |
| `sep.join(iterable)` | `"-".join(["2025","10","16"])` | `'2025-10-16'` | Une con separador |

## Recorte y alineación

| Método | Uso | Resultado | Notas |
|---|---|---|---|
| `strip(chars=None)` | `"  hola  ".strip()` | `'hola'` | Extremos |
| `lstrip(chars=None)` | `"--hola".lstrip("-")` | `'hola'` | Izquierda |
| `rstrip(chars=None)` | `"hola--".rstrip("-")` | `'hola'` | Derecha |
| `center(width[, fill])` | `"hi".center(6,"*")` | `'**hi**'` | Centrar |
| `ljust(width[, fill])` | `"hi".ljust(5,".")` | `'hi...'` | Izquierda |
| `rjust(width[, fill])` | `"hi".rjust(5,".")` | `'...hi'` | Derecha |
| `zfill(width)` | `"42".zfill(5)` | `'00042'` | Ceros a la izquierda |

## Partición (en 3 partes)

| Método | Uso | Resultado | Notas |
|---|---|---|---|
| `partition(sep)` | `"a=b=c".partition("=")` | `('a', '=', 'b=c')` | Desde la izquierda |
| `rpartition(sep)` | `"a=b=c".rpartition("=")` | `('a=b', '=', 'c')` | Desde la derecha |

## Formato de cadenas  
| Técnica | Uso | Resultado | Notas |
|---|---|---|---|
| f-strings | `f"Hola {2+3}"` | `'Hola 5'` | Clara y rápida |
| `format` | `"{x}+{y}={z}".format(x=2,y=3,z=5)` | `'2+3=5'` | Flexible |
| `format_map` | `"{x} {y}".format_map({'x':1,'y':2})` | `'1 2'` | Con diccionario |

## Comprobaciones (booleanas)

| Método | ¿Qué verifica? | Ejemplo | Resultado |
|---|---|---|---|
| `isalpha()` | Solo letras | `"Hola".isalpha()` | `True` |
| `isdigit()` | Dígitos Unicode | `"²3".isdigit()` | `True` (superíndice ²) |
| `isdecimal()` | Dígitos 0–9 | `"23".isdecimal()` | `True` |
| `isnumeric()` | Números (romanos, fracciones, etc.) | `"Ⅷ".isnumeric()` | `True` |
| `isalnum()` | Letras o dígitos | `"abc123".isalnum()` | `True` |
| `islower()` / `isupper()` / `istitle()` | Caso del texto | `"Hola".istitle()` | `True` |
| `isspace()` | Solo espacios en blanco | `"\t\n".isspace()` | `True` |
| `isidentifier()` | Identificador Python válido | `"_x1".isidentifier()` | `True` |
| `isascii()` | Solo ASCII | `"á".isascii()` | `False` |
| `isprintable()` | Imprimible | `"Hola".isprintable()` | `True` |

## Otros útiles

| Método | Uso | Resultado | Notas |
|---|---|---|---|
| `removeprefix(pref)` | `"archivo".removeprefix("ar")` | `'chivo'` | 3.9+ |
| `removesuffix(suf)` | `"nota.txt".removesuffix(".txt")` | `'nota'` | 3.9+ |
| `encode(encoding)` | `"á".encode("utf-8")` | `b'\xc3\xa1'` | A bytes |
| `decode(encoding)` | `b'\xc3\xa1'.decode("utf-8")` | `'á'` | A str |
