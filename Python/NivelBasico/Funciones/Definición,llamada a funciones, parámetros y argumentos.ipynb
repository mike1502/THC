#¿Qué es una función?

En cálculo, una función como

$$f(x) = x^2 + 1$$

es una regla de correspondencia pura: toma un valor de entrada, le aplica una transformación matemática y devuelve siempre el mismo resultado, sin alterar nada fuera de ella.

En Python, una función es mucho más amplia: es un bloque de código reutilizable con nombre propio que encapsula una tarea específica. Puede comportarse como una función matemática (recibir datos y devolver un resultado con return), pero también puede ejecutarse sin recibir nada, sin devolver nada (None), o provocando efectos secundarios (side effects), como imprimir en pantalla, guardar un archivo o modificar datos externos.

#Variables locales y globales

Al ser un bloque de ejecución independiente, cada vez que se llama a una función en Python se crea un entorno privado en memoria exclusivo para ella. A esta frontera de visibilidad se le llama ámbito (scope), y es lo que divide las variables en dos tipos:

- Variable local: Se define dentro de la función (incluidos sus parámetros). Nace cuando la función empieza a ejecutarse y se destruye en cuanto termina. Nadie fuera de esa función puede verla ni usarla.

- Variable global: Se define en el cuerpo principal del archivo (fuera de cualquier def). Vive durante toda la ejecución del programa y cualquier función puede leerla, pero ninguna puede modificarla directamente por accidente.

# Variables globales
[regresar al contenido](#c1)

Las variables que se crean fuera de una función se conocen como variables globales.

Las variables globales pueden ser utilizadas por todos, tanto dentro de funciones como fuera de ellas.

````python
x = "awesome"

def myfunc():
  print("Python is " + x)

myfunc()
````

Si creas una variable con el mismo nombre dentro de una función, esta variable será **local** y solo se podrá usar dentro de la función. La variable global con el mismo nombre permanecerá como estaba, global y con el valor original.

````python
x = "awesome"

def myfunc():
  x = "fantastic"
  #print("Python is " + x)

myfunc()

print("Python is " + x)
````

## Variables locales y la palabra clave *global*
[regresar al contenido](#c1)

Normalmente, cuando se crea una variable dentro de una función, esa variable es local y solo se puede usar dentro de esa función.

Para crear una variable global dentro de una función, puedes usar la palabra clave ***global***.

````python
x = "awesome"
def myfunc():
    global x  #la defino como variable global
    x = "fantastic"
    print("Python is " + x)

myfunc()

print("Python is " + x)
````

Si usas la palabra clave **global**, la variable pertenece al ámbito global.

## *args
[regresar al contenido](#c1)

Si no sabes cuántos argumentos se pasarán a tu función, agrega un * antes del nombre del parámetro en la definición de la función.

De esta manera, la función recibirá una **tupla** de argumentos y podrá acceder a los elementos correspondientes.

````python
def mi_funcion(*kids):
    print("El niño menor es " + kids[2])

mi_funcion("Emil", "Tobias", "Linus")
````

<a name='4-2'></a>

## Clave Arguments
[regresar al contenido](#c1)

También puedes enviar argumentos utilizando la sintaxis clave = valor.

De esta manera, el orden de los argumentos no importa.

````python
def mi_funcion(niño3,niño2,niño1):
    print("El niño menor es " + niño3)

mi_funcion(niño1 = "Emil", niño2 = "Tobias", niño3 = "Linus")
````

<a name='4-3'></a>

## **kwargs
[regresar al contenido](#c1)

Si no sabes cuántos argumentos de palabras clave se pasarán a tu función, agrega dos asteriscos: ** antes del nombre del parámetro en la definición de la función.

De esta manera, la función recibirá un diccionario de argumentos y podrá acceder a los elementos correspondientes.

````python
def mi_funcion(**kid):
  print("El apellido es " + kid["Ape"])

mi_funcion(nombre = "Juan", Ape = "Pérez")
````

# Pasando una estructura de datos como parámetros
[regresar al contenido](#c1)

Puedes enviar cualquier tipo de dato como argumento a una función (cadena, número, lista, diccionario, etc.), y se tratará como el mismo tipo de dato dentro de la función.

Por ejemplo, si envías una lista como argumento, seguirá siendo una lista cuando llegue a la función.

````python
def mi_funcion(frutas):
  for x in frutas:
    print(x)

frutas = ["manzana", "banana", "mora"]

mi_funcion(frutas)
````

# Un ejemplo numérico

El comando **return** sirve para devolver un resultado desde una función al lugar donde fue llamada.

````python
#Calcular el área de un triángulo
def area_triangulo(base, altura):
    return (base * altura) / 2

# Imprime el resultado en pantalla e ingresa los valores como argumentos (base y altura).
print("Área del triángulo:", area_triangulo(10, 5))
````

````python
##O bien se puede hacer de esta forma, solicitando los datos al usuario.
base = float(input("Ingresa la base del triángulo: "))
altura = float(input("Ingresa la altura del triángulo: "))

# Llamar a la función
area = area_triangulo(base, altura)

print("El área del triángulo es:", area)
````
