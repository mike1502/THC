# 1. ¿Qué es SymPy? (exactitud vs aproximación)
[regresar al contenido](#c1)

**SymPy** es una biblioteca de Python para **matemáticas simbólicas**.  
A diferencia de `numpy` (que se enfoca en números), SymPy trabaja con expresiones **exactas**.

Ejemplos rápidos de lo que significa “exacto”:

- \(\sqrt{2}\) se guarda como \(\sqrt{2}\), no como `1.4142...`
- \(\pi\) se guarda como \(\pi\), no como `3.14159...`
- \(\frac{1}{3}\) se guarda como fracción exacta, no como `0.3333...`

Vamos a comparar simbólico vs numérico.

```
import sympy as sp
sp.init_printing()  # imprime bonito en notebooks

# Comparación simbólico vs numérico
expr = sp.sqrt(2) + sp.Rational(1, 3) + sp.pi
expr
```

```
# Aproximación numérica (float)
expr.evalf()
```

**Idea clave:**

- El mundo *simbólico* sirve para **manipular** expresiones (factorizar, simplificar, derivar, integrar, etc.).
- El mundo *numérico* sirve para **evaluar** rápido en muchos puntos o para aproximaciones.

SymPy te deja moverte entre ambos: simbólico → `evalf()`.

# 3. Símbolos, expresiones y evaluación
[regresar al contenido](#c1)

En SymPy, primero defines **símbolos** (variables) y luego construyes expresiones.

- `sp.Symbol('x')` crea un símbolo individual.
- `sp.symbols('x y z')` crea varios símbolos de forma rápida.

```
x = sp.Symbol('x')
y, z = sp.symbols('y z')
x, y, z
```

Construyamos una expresión y veamos sus partes:

```
expr = (x**2 + 2*x + 1) / (x + 1)
expr
```

```
# Partes útiles para inspección:
expr.free_symbols, expr.func, expr.args
```

Evaluación numérica (por sustitución):

- `subs({x: 2})` sustituye \(x=2\) simbólicamente.
- `evalf()` aproxima numéricamente.

```
expr_sub = expr.subs({x: 2})
expr_sub, expr_sub.evalf()
```

# 4. Simplificar, expandir y factorizar
[regresar al contenido](#c1)

Algunas funciones clave:

- `sp.simplify(expr)` intenta simplificar “en general”.
- `sp.expand(expr)` expande productos y potencias.
- `sp.factor(expr)` factoriza.
- `sp.cancel(expr)` cancela factores en fracciones racionales.

```
expr = (x**2 + 2*x + 1) / (x + 1)
sp.simplify(expr)
```

```
expr2 = (x - 1)*(x + 2) - (x**2 + x - 2)
sp.expand(expr2), sp.simplify(expr2)
```

```
expr3 = x**3 - 3*x**2 - 4*x + 12
sp.factor(expr3)
```

```
expr4 = (x**2 - 1)/(x - 1)
sp.cancel(expr4), sp.simplify(expr4)
```

# 5. Sustitución y evaluación numérica
[regresar al contenido](#c1)

Sustitución con `subs`:

- Sustituir un valor: `expr.subs(x, 1)`
- Sustituir varios: `expr.subs({x: 1, y: 2})`
- Sustituir por otra expresión: `expr.subs(x, y+1)`

También puedes obtener una función numérica rápida con `lambdify`.

```
expr = sp.sin(x) + sp.exp(-x)
expr.subs(x, 0), expr.subs(x, sp.pi/2)
```

```
# Función numérica (para arrays, gráficas, etc.)
import numpy as np

f_num = sp.lambdify(x, expr, modules=["numpy"])
xs = np.linspace(0, 5, 6)
f_num(xs)
```

# 6. Derivadas (una y varias variables)
[regresar al contenido](#c1)

Derivadas con `sp.diff`:

- `sp.diff(f, x)` derivada de \(f\) respecto a \(x\)
- `sp.diff(f, x, 2)` segunda derivada
- `sp.diff(f, x, y)` derivada parcial \(\partial^2 f / (\partial y\,\partial x)\)

También puedes usar el método `.diff(...)` en la expresión.

```
f = sp.sin(x)*sp.exp(x)
sp.diff(f, x)
```

```
sp.diff(f, x, 2)
```

```
x, y = sp.symbols('x y')
g = x**2*y + sp.sin(x*y)
dg_dx = sp.diff(g, x)
dg_dy = sp.diff(g, y)
dg_dx, dg_dy
```

# 7. Integrales (indefinidas y definidas)
[regresar al contenido](#c1)

Integrales con `sp.integrate`:

- `sp.integrate(f, x)` integral indefinida \(\int f\,dx\)
- `sp.integrate(f, (x, a, b))` integral definida \(\int_a^b f\,dx\)

Cuando SymPy no puede resolver una integral, a veces devuelve la integral “sin resolver” (lo cual también es útil).

```
x = sp.Symbol('x')
f = sp.cos(x)**2
F = sp.integrate(f, x)
F
```

```
# Integral definida
sp.integrate(sp.exp(-x), (x, 0, sp.oo))
```

# 8. Límites y series
[regresar al contenido](#c1)

- `sp.limit(expr, x, a)` calcula \(\lim_{x\to a}\)
- `sp.series(expr, x, a, n)` serie de Taylor alrededor de \(a\) hasta orden \(n-1\)

Veamos ejemplos clásicos.

```
x = sp.Symbol('x')
sp.limit(sp.sin(x)/x, x, 0)
```

```
# Serie de Taylor de e^x alrededor de 0
sp.series(sp.exp(x), x, 0, 6)
```

```
# Quitar el término O(x^n) para quedarte con un polinomio
sp.series(sp.exp(x), x, 0, 6).removeO()
```

# 9. Resolver ecuaciones y sistemas
[regresar al contenido](#c1)

Funciones típicas:

- `sp.solve(eq, x)` resuelve ecuaciones algebraicas (exacto si se puede).
- `sp.solve([eq1, eq2], [x, y])` resuelve sistemas.
- `sp.nsolve(...)` resuelve numéricamente (cuando quieres aproximaciones).

Ojo: a veces SymPy devuelve varias soluciones (listas).

```
x = sp.Symbol('x')
sol = sp.solve(sp.Eq(x**2 - 5*x + 6, 0), x)
sol
```

```
x, y = sp.symbols('x y')
eq1 = sp.Eq(2*x + y, 7)
eq2 = sp.Eq(x - y, 1)
sp.solve([eq1, eq2], [x, y])
```

```
# nsolve: solución numérica (necesita un punto inicial)
x = sp.Symbol('x')
sp.nsolve(sp.cos(x) - x, 0.7)
```

# 10. Álgebra lineal simbólica (matrices)
[regresar al contenido](#c1)

SymPy incluye matrices simbólicas (`sp.Matrix`) con operaciones típicas:

- determinante: `.det()`
- inversa: `.inv()` (si existe)
- eigenvalores: `.eigenvals()`
- resolver sistemas lineales: `A.LUsolve(b)`

Esto es muy útil cuando quieres expresiones exactas.

```
a, b, c = sp.symbols('a b c')
A = sp.Matrix([[1, a], [b, c]])
A, A.det()
```

```
# Sistema lineal A*[x,y]^T = b
x, y = sp.symbols('x y')
A = sp.Matrix([[2, 1], [1, -1]])
bvec = sp.Matrix([7, 1])
sol = A.LUsolve(bvec)
sol
```

# 11. Funciones y ecuaciones diferenciales (mini-intro)
[regresar al contenido](#c1)

SymPy puede trabajar con funciones simbólicas y resolver ecuaciones diferenciales sencillas.

- `sp.Function('y')` define una función \(y(x)\).
- `sp.dsolve(...)` intenta resolver una ecuación diferencial ordinaria (EDO).

Ejemplo: \(y' = y\) con solución \(y = C e^x\).

```
x = sp.Symbol('x')
y = sp.Function('y')

edo = sp.Eq(sp.diff(y(x), x), y(x))
sp.dsolve(edo)
```
