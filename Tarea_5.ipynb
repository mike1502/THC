**Ejercicio 7.- Cálculo de la distancia entre dos puntos mediante la fórmula de Haversine.**

La fórmula de Haversine permite calcular la distancia entre dos puntos sobre la superficie de la Tierra, a partir de sus coordenadas geográficas. La fórmula está dada por la siguiente expresión:

\begin{equation}
d = 2R  \arcsin \left( \sqrt{\mathrm{haversin}(\varphi_2 - \varphi_1) + \cos(\varphi_1)\cos(\varphi_2)\mathrm{haversin}(\lambda_2 - \lambda_1)} \right)
\end{equation}

donde:

d = distancia entre dos puntos (en km),

R = radio de la Tierra (𝑅≈6378.1),

𝜑1, 𝜑2 = latitudes de los dos puntos.

𝜆1, 𝜆2 = longitudes de los dos puntos (en radianes).

La función de Haversine se define como:

 \begin{equation}
\mathrm{haversin}(\alpha) = \sin^2 \left(\frac{\alpha}{2}\right)
\end{equation}

Para convertir de grados a radianes se utiliza la relación:

\begin{equation}
\text{radianes} = \text{grados} \times \frac{\pi}{180}
\end{equation}
