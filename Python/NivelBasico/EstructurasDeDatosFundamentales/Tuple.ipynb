# Tuplas
[regresar al contenido](#c1)


Una tupla es una colección ordenada e inmutable de elementos. A diferencia de las listas, las tuplas no pueden modificarse después de su creación. Se definen utilizando paréntesis y pueden contener diferentes tipos de elementos, como números, cadenas, booleanos, e incluso otras tuplas. Las tuplas son útiles cuando necesitas almacenar un conjunto de valores que no deben cambiar.

## "Borrar" elementos en una tupla
[regresar al contenido](#c1)

Si desea borrar el valor de la lista en un índice específico se hace de la siguiente manera.

En ese código no se borra el elemento de la tupla original; se crea una tupla completamente nueva en otro espacio de memoria y dejando la original intacta.

Para comprobarlo, observa qué pasa si imprimes ambas variables y revisas sus direcciones en memoria con la función id():

```python
mi_tupla = (1, 2, 3, 4, 5)
elemento_a_borrar = 3

# Aquí se construye una nueva tupla desde cero
nueva_tupla = tuple(elemento for elemento in mi_tupla if elemento != elemento_a_borrar)

print("Original:", mi_tupla, "-> ID en memoria:", id(mi_tupla))
print("Nueva:   ", nueva_tupla, "-> ID en memoria:", id(nueva_tupla))
```

## Agregar elementos
[regresar al contenido](#c1)

Para agregar una nueva variable se realiza de la siguiente manera.
Observe que no se manipula directamente la tupla se utiliza con la función
**tuple()**. Se trata de una función integrada (o constructor de clase) de Python. Su trabajo es tomar cualquier objeto iterable (como una lista, una cadena de texto, un conjunto set o un rango) y construir una tupla nueva a partir de sus elementos.

```python
mi_lista = [1, 2, 3]
mi_lista.append(4)
mi_lista.append(5)
mi_lista=tuple(mi_lista)
print(mi_lista)  # Resultado: [1, 2, 3, 4, 5]
```

# Ejemplos
[regresar al contenido](#c1)


Al igual que "borrar", no puedes agregar elementos directamente a una tupla existente (no tienen métodos como .append() o .insert()). La única forma es construir una tupla nueva combinando la original con los nuevos datos. Existen cuatro formas de agregar elementos.

1.- Concatenación con el operador + (La más directa)
Sumas la tupla original con otra tupla que contenga los nuevos elementos.

```python
mi_tupla = (1, 2, 3)

# ¡OJO! Para un solo elemento, la coma final es obligatoria para que Python sepa que es una tupla
mi_tupla = mi_tupla + (4,)

# Para varios elementos:
mi_tupla = mi_tupla + (5, 6)

print(mi_tupla)  # Salida: (1, 2, 3, 4, 5, 6)
```

2. Desempaquetado con * (Más moderna y limpia)
Disponible desde Python 3.5, permite "abrir" los elementos de la tupla original dentro de una nueva, ideal si quieres insertar elementos al principio, en medio o al final.

```python
mi_tupla = (2, 3)

# Agregamos al inicio y al final en un solo paso
nueva_tupla = (1, *mi_tupla, 4, 5)

print(nueva_tupla)  # Salida: (1, 2, 3, 4, 5)
```

3. Conversión temporal a lista (Útil para cambios complejos)
Si necesitas hacer varias inserciones en posiciones específicas o modificaciones seguidas, lo más práctico y eficiente es convertirla a lista, modificarla y volverla a convertir a tupla.

```python
mi_tupla = ("a", "c", "d")

lista_temp = list(mi_tupla)
lista_temp.insert(1, "b")  # Insertamos en el índice 1
lista_temp.append("e")     # Agregamos al final

mi_tupla = tuple(lista_temp)
print(mi_tupla)  # Salida: ('a', 'b', 'c', 'd', 'e')
```

4. Operador de asignación +=
Es un atajo visual del punto 1. Aunque parece que la estás modificando in-place, por debajo Python sigue creando un objeto nuevo en memoria y reasignando la variable:

```python
mi_tupla = (10, 20)
mi_tupla += (30,)

print(mi_tupla)  # Salida: (10, 20, 30)
```

# Utilidad de las tuplas

[regresar al contenido](#c1)

Si se utilizan listas (punto1 = [3, 5]), el cálculo matemático funcionaría
igual, pero se pierde el hecho de que un punto siempre tenga 2 dimensiones
inalterables y el código sería más propenso a errores en un proyecto grande.


```python
# Definir tuplas para coordenadas
punto1 = (3, 5)
punto2 = (-2, 7)
punto3 = (0, 0)

# Acceder a los valores de las coordenadas
x1, y1 = punto1
x2, y2 = punto2
x3, y3 = punto3

# Calcular la distancia entre dos puntos
def distancia(p1, p2):
    x1, y1 = p1
    x2, y2 = p2
    return ((x2 - x1)**2 + (y2 - y1)**2)**0.5

distancia_entre_puntos = distancia(punto1, punto2)
print("La distancia entre punto1 y punto2 es:", distancia_entre_puntos)
```

## Importante

Aunque una tupla es inmutable, si contiene un objeto mutable en su interior (como una lista), el contenido de esa lista interna sí puede cambiar:

```python
tupla_mixta = ("Python", [10, 20])
tupla_mixta[1].append(30)  # Esto funciona porque la referencia a la lista no cambia, solo su contenido
print(tupla_mixta)  # Salida: ('Python', [10, 20, 30])
```

## Ejemplo con ciclo for

```python
# Recorre la tupla
dias_finde = ("Sábado", "Domingo")

for dia in dias_finde:
    print(f"Hoy es {dia}, toca descansar.")
```

```python
# Recorriendo y desempaquetar una tupla de tuplas
productos = (("Laptop", 1200), ("Mouse", 25), ("Teclado", 80))

for nombre, precio in productos:
    print(f"El producto {nombre} cuesta ${precio}")
```

## Ejemplo con if-else

```python
usuarios_vip = ("Ana", "Carlos", "Elena")
usuario_actual = "Carlos"

# Comprobamos pertenencia en la tupla
if usuario_actual in usuarios_vip:
    print(f"Bienvenido {usuario_actual}, tienes acceso total.")
else:
    print("Acceso estándar.")
```

## Ejercicio conbinado for/if-else

```python
valores = (4, 12, 7, 25, 10)  # Tupla de valores
rango = (5, 15)               # Tupla con máximo y mínimo

inicio, fin = rango  # Desempaquetamos los límites

for numero in valores:
    if inicio <= numero <= fin:
        print(f"El {numero} está DENTRO del rango.")
    else:
        print(f"El {numero} está FUERA del rango.")

enumerated_values = list(enumerate(valores))
print("---------------")
print(enumerated_values) #para identificar el índice de la tupla y su valor
```

## Funciones más utilizadas para las tuplas

| Función / Operador | Descripción | Ejemplo | Resultado |
| :--- | :--- | :--- | :--- |
| `len(t)` | Devuelve la cantidad total de elementos en la tupla. | `len((10, 5, 20))` | `3` |
| `max(t)` | Devuelve el elemento con el valor máximo (requiere datos comparables). | `max((10, 5, 20))` | `20` |
| `min(t)` | Devuelve el elemento con el valor mínimo. | `min((10, 5, 20))` | `5` |
| `sum(t)` | Suma todos los elementos (solo si son numéricos). | `sum((10, 5, 20))` | `35` |
| `sorted(t)` | Devuelve una **lista** nueva con los elementos ordenados de menor a mayor. | `sorted((10, 5, 20))` | `[5, 10, 20]` |
| `any(t)` | Devuelve `True` si **al menos un** elemento es verdadero (distinto de `0`, `False` o vacío). | `any((0, False, 5))` | `True` |
| `all(t)` | Devuelve `True` si **todos** los elementos son verdaderos. | `all((10, 5, 20))` | `True` |
| `enumerate(t)` | Devuelve un iterador que empareja cada índice con su valor `(índice, valor)`. | `list(enumerate(('a', 'b')))` | `[(0, 'a'), (1, 'b')]` |
| `zip(t1, t2)` | Combina dos o más tuplas elemento por elemento en pares. | `list(zip((1, 2), ('a', 'b')))` | `[(1, 'a'), (2, 'b')]` |
| `in` / `not in` | Operador de pertenencia: comprueba si un valor existe dentro de la tupla. | `5 in (10, 5, 20)` | `True` |
